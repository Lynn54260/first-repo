```python
import dlt
from pyspark.sql.functions import *

@dlt.table(
    name="bronze_sales",
    comment="Raw sales data from CSV"
)
def bronze_sales():

    return (
        spark.read.format('csv')
        .option("cloudFiles.format", "csv")
        .option("cloudFiles.format", "csv")
        .option("header", "true")
        .load("/FileStore/sales_input")
    )
```
